```python
# Example 2: Order of Operations (PEMDAS)
result1 = 10 + 5 * 2
result2 = (10 + 5) * 2

print("Order of Operations:")
print(f"10 + 5 * 2 = {result1}  (Multiplication first)")
print(f"(10 + 5) * 2 = {result2}  (Parentheses first)")
```
