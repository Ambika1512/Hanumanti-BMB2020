```python
def analyze_farming_practice(practice, invest, yield_amt, price, cost):
    revenue = yield_amt * price
    profit = revenue - cost - invest
    return {"Practice": practice, "Profit": profit}

print(analyze_farming_practice("Organic", 70000, 35, 10000, 75000))
def recommend_next_crop(current, n, p, season):
    if current == "Wheat":
        return "Grow Pulses next"
    else:
        return "Grow Wheat next"

print(recommend_next_crop("Wheat", 80, 20, "Rabi"))
def assess_pest_risk(temp, hum, age, variety):
    if temp > 30 and hum > 70:
        return "HIGH"
    return "LOW"

print(assess_pest_risk(32, 75, 20, 4))
```

```python
def identify_disease(color, pattern, part):
    if color == "yellow" and part == "leaves":
        return "Possible leaf blight"
    return "Unknown disease"

print(identify_disease("yellow", "scattered", "leaves"))
def is_ready(days, moisture):
    return days > 120 and moisture < 20

print(is_ready(130, 18))
def soil_amendment(ph):
    if ph < 6:
        return "Add lime"
    else:
        return "Soil OK"

print(soil_amendment(5.5))
def weather_impact(temp, rain):
    if rain < 50:
        return "Low yield expected"
    return "Good yield expected"

print(weather_impact(28, 120))
```

```python
seeds = {
    "Wheat": {"quantity": 500, "expiry": "2026-06", "price": 45},
    "Rice": {"quantity": 300, "expiry": "2026-03", "price": 55}
}

# Add new seed
seeds["Cotton"] = {"quantity": 200, "expiry": "2026-12", "price": 120}

# Update quantity
seeds["Wheat"]["quantity"] -= 50

# Inventory value
total = 0
for s in seeds:
    total += seeds[s]["quantity"] * seeds[s]["price"]

print("Inventory Value:", total)
```
