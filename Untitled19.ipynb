```python
import pandas as pd
import numpy as np


farm_data={
    "crop":["wheat","rice","millet","pulses"],
    "yield_kg":[3200,np.nan,3500,3000],
    "water_liters":[4200,5500,np.nan,3000],
     "cost_inr":[1800,2000,1700,np.nan]

}


df=pd.DataFrame(farm_data)
df
```

```python
from sklearn.linear_model import LinearRegression

x=df[["water_liters"]].dropna()
y=df["yield_kg"].dropna()

# Align indices after dropping NaNs
aligned_x, aligned_y = x.align(y, join='inner', axis=0)

#very simple model
model_simple=LinearRegression()
model_simple.fit(aligned_x,aligned_y)

# Predict using a value for 'water_liters'
# Note: You might need to provide a value within the trained range or consider scaling
model_simple.predict([[4000]])
```

```python
from sklearn import pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression

#very complex model
model_complex=make_pipeline(
    PolynomialFeatures(degree=5),
    LinearRegression()
)

# Use aligned_x and aligned_y from the previous cell
model_complex.fit(aligned_x,aligned_y)

# Predict using a value for 'water_liters'
# Note: You might need to provide a value within the trained range or consider scaling
model_complex.predict([[205]])
```
