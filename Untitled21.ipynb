```python
import pandas as pd
#crop health dataset
data={
    "yellow leaves":[1,0,1,0,1,1,0,0],
    "spots_present":[1,0,0,1,0,0,0,0],
    "disease":["fungal", "nutrient","healthy",
               "healthy","fungal","healthy",
               "nutrient","healthy"]
}

df=pd.DataFrame(data)
df
```

```python
from sklearn.tree import DecisionTreeClassifier

x=df[["yellow leaves","spots_present"]]#symptoms
y=df["disease"]#diagonosis

model=DecisionTreeClassifier(random_state=42)
model.fit(x,y)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plot_tree(model,
          feature_names=x.columns,
          class_names=model.classes_,
          filled=True)
plt.show()
```

```python
import pandas as pd

feature_importance = pd.Series(model.feature_importances_, index=x.columns)
feature_importance.sort_values(ascending=False).plot(
    kind="barh",
    title="Feature Importance"
)
plt.show()
```
